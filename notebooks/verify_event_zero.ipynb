# Event Zero: reproduce the every-family verification in one click

This notebook clones Event Zero, builds the synthetic verification corpus, runs **The Verdict** over every fixture, and prints the same results table as the GitHub Actions verification workflow.

Safety by construction: every fixture is synthetic and disarmed (proof atoms as data, `.invalid` and TEST-NET addresses only, headers neutralised where applicable). Fixtures are analysed statically and **never executed**. Nothing in the corpus mines, cracks, or attacks.

```python
# 1. Get the project (Colab: clone; locally: point at your checkout)
import os, subprocess, sys
if not os.path.isdir('Event_Zero'):
    subprocess.run(['git', 'clone', 'https://github.com/Epsilon0xCell/Event_Zero.git'], check=True)
os.chdir('Event_Zero')
print(os.getcwd())
```

```python
# 2. Install dependencies
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-r', 'requirements.txt', 'unicorn', 'cryptography'], check=True)
print('installed')
```

```python
# 3. Build the synthetic corpus (37 entries: 20 families, FP-twin controls, designed-OPEN cases, surprise fixtures)
subprocess.run([sys.executable, 'fixtures/verification/build_corpus.py'], check=True)
```

```python
# 4. Run The Verdict over the whole corpus, with the independent scorer and the determinism double-run
r = subprocess.run([sys.executable, 'fixtures/verification/run_verification.py'], capture_output=True, text=True)
print(r.stdout[-6000:])
print('exit code:', r.returncode, '(0 means PASS)')
```

```python
# 5. Show the results table
print(open('fixtures/verification/results_table.md').read())
```
